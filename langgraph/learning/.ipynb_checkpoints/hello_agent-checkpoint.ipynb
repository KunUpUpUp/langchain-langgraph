```python
from langgraph.graph import StateGraph
from typing import TypedDict
from IPython.display import Image, display


class AgentState(TypedDict):
    message: str


def hello(state: AgentState) -> AgentState:
    state["message"] = "Hey" + state["message"]
    return state

def bye(state: AgentState) -> AgentState:
    state["message"] = "Bye" + state["message"]
    return state

def build_graph():
    graph = StateGraph(AgentState)
    graph.add_node("hello", hello)
    graph.add_node("bye", bye)
    graph.set_entry_point("hello")
    # graph.add_edge(START, "hello")
    graph.add_edge("hello", "bye")
    graph.set_finish_point("bye")
    # graph.add_edge("bye", END)
    graph_compile = graph.compile()
    return graph_compile

def display_graph(graph):
    display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
graph = build_graph()
display_graph(graph)
```

```python
state = AgentState(message="Jude")
result = graph.invoke(state)
result["message"]
```
